```python
#import necessary functions
from pyspark.sql.functions import *
from pyspark.sql.types import *

# create the silver schema
spark.sql("""
CREATE SCHEMA IF NOT EXISTS playerforge.silver
""")

playerforge_silver_table = "playerforge.silver.game_stats"

silver_df = spark.read.table("playerforge.bronze.game_stats")

null_dict = {
    "team_name": "Unknown",
    "player_role": "Unknown",
    "map_played": "Unknown",
    "match_type": "Unknown",
    "match_outcome": "Unknown",
    "mvp_award": "Unknown"
}

silver_df = silver_df.fillna(null_dict)

(
    silver_df.write.
    format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(playerforge_silver_table)
)
display(silver_df)
```
