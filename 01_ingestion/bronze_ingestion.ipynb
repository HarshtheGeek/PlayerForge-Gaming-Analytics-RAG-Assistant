```python
# imported the necessary libraries
from pyspark.sql.functions import *
from pyspark.sql.types import *

# create the bronze schema
spark.sql("""
CREATE SCHEMA IF NOT EXISTS playerforge.bronze
""")

# path to the raw data
raw_data_path = "/Volumes/playerforge/playerforge_raw/game_data"

# create the bronze table
playerforge_bronze_table = "playerforge.bronze.game_stats"

# reading the data
bronze_file_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("InferSchema","true")
    .load(raw_data_path)
)
bronze_file_df = bronze_file_df.withColumn("_ingested_date", current_timestamp())
display(bronze_file_df)

(
    bronze_file_df.write
    .format("delta")
    .mode("append")
    .saveAsTable(playerforge_bronze_table)
)
```
