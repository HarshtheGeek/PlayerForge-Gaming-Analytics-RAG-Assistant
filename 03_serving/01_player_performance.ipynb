```sql
CREATE OR REPLACE TABLE playerforge.gold.player_stats AS
SELECT
    player_id,
    COUNT(*) AS matches_played,
    SUM(kills) AS total_kills,
    SUM(assists) AS total_assists,
    SUM(deaths) AS total_deaths,
    ROUND(AVG(accuracy_percent),2) AS avg_accuracy,
    ROUND(AVG(reaction_time_ms),2) AS avg_reaction_time,
    ROUND(AVG(fatigue_index),2) AS avg_fatigue,
    ROUND(AVG(performance_score),2) AS avg_performance_score
FROM playerforge.silver.game_stats
GROUP BY player_id;
```
