```sql
-- ==========================================================

-- Player Performance RAG Analysis

-- ==========================================================
-- Question: Which players demonstrate the strongest overall performance, and what factors including combat efficiency, win rate, accuracy, reaction time, fatigue, role, map performance, and MVP recognition contribute to their performance?

-- Metrics:
-- player_id - 
-- player_role
-- matches_played - 
-- total_kills - 
-- total_assists - 
-- total_deaths - 
-- kd_ratio - 
-- kills_per_match - 
-- assists_per_match - 
-- win_rate
-- avg_win_probability
-- avg_accuracy
-- avg_reaction_time
-- avg_fatigue
-- avg_performance_score
-- mvp_count
-- mvp_rate
-- best_map
-- best_map_performance

-- ==========================================================

SELECT * FROM playerForge.silver.game_stats;
DESCRIBE playerForge.silver.game_stats;

CREATE OR REPLACE TABLE playerForge.gold.player_performance AS
SELECT player_id,
COUNT(*) AS matches_played,
SUM(kills) AS total_kills,
SUM(assists) AS total_assists,
SUM(deaths) AS total_deaths,
ROUND(SUM(kills) / NULLIF(SUM(deaths),0),2) AS kd_ratio,
ROUND(SUM(kills) / COUNT(*),2) as kills_per_match,
ROUND(SUM(assists) / COUNT(*),2) as assists_per_match,
ROUND(
    SUM(CASE WHEN match_outcome = 'Win' THEN 1 ELSE 0 END)
    * 100.0 / COUNT(*),
    2
) AS win_rate
FROM playerForge.silver.game_stats
GROUP BY player_id ; 
```
