```sql
-- ==========================================================
-- Win Rate Analysis
-- ==========================================================
-- Question:
-- Which players have the highest win rates and how does their
-- win probability compare with their actual match outcomes?
--
-- Metrics:
-- matches_played
-- wins
-- losses
-- win_rate
-- avg_win_probability
-- ==========================================================

CREATE OR REPLACE TABLE playerforge.gold.win_rate_stats AS

SELECT
    player_id,

    COUNT(*) AS matches_played,

    SUM(CASE
        WHEN match_outcome = 'Win' THEN 1
        ELSE 0
    END) AS wins,

    SUM(CASE
        WHEN match_outcome = 'Loss' THEN 1
        ELSE 0
    END) AS losses,

    ROUND(
        SUM(CASE
            WHEN match_outcome = 'Win' THEN 1
            ELSE 0
        END) * 100.0 / NULLIF(COUNT(*), 0),
        2
    ) AS win_rate,

    ROUND(AVG(win_probability), 2) AS avg_win_probability

FROM playerforge.silver.game_stats

GROUP BY player_id

ORDER BY win_rate DESC;
```
