```sql
-- ==========================================================
-- MVP Performance Analysis
-- ==========================================================
-- Question:
-- Which players receive the most MVP awards and what
-- performance characteristics are associated with MVP recognition?
--
-- Metrics:
-- mvp_count
-- mvp_rate
-- avg_performance_score
-- avg_kills
-- avg_assists
-- avg_accuracy
-- avg_reaction_time
-- ==========================================================
CREATE OR REPLACE TABLE playerforge.gold.mvp_stats AS 
WITH role_metrics AS (
    SELECT
    map_played,
    COUNT(player_id) AS players,
    COUNT(match_type) AS matches,
    ROUND(AVG(kills),2) AS avg_kills,
    ROUND(AVG(assists),2) AS avg_assists,
    ROUND(AVG(deaths),2) AS avg_deaths,
    ROUND(AVG(reaction_time_ms),2) AS avg_reaction_time,
    ROUND(AVG(performance_score),2) AS avg_performance_score
    FROM playerforge.silver.game_stats
    GROUP BY map_played
)
SELECT * FROM role_metrics;
SELECT * FROM playerforge.silver.game_stats; 
```
