```sql
-- ==========================================================
-- Combat Performance Analysis
-- ==========================================================
-- Question:
-- Which players demonstrate the strongest combat efficiency
-- based on their K/D ratio, kills per match, and assists per match?
--
-- Metrics:
-- kd_ratio
-- kills_per_match
-- assists_per_match
-- ==========================================================

CREATE OR REPLACE TABLE playerforge.gold.combat_stats AS

WITH kd_metrics AS (
    SELECT
        player_id,
        SUM(kills) AS total_kills,
        SUM(deaths) AS total_deaths,
        SUM(assists) AS total_assists,
        COUNT(*) AS matches_played
    FROM playerforge.silver.game_stats
    GROUP BY player_id
)

SELECT
    player_id,
    ROUND(total_kills / NULLIF(total_deaths, 0), 2) AS kd_ratio,
    ROUND(total_kills / NULLIF(matches_played, 0), 2) AS kills_per_match,
    ROUND(total_assists / NULLIF(matches_played, 0), 2) AS assists_per_match
FROM kd_metrics
ORDER BY kd_ratio DESC;
```
